```python
# Program: Variables and Data Types in Python

# Integer data type
age = 20
print("Age:", age)
print("Type of age:", type(age))

# Float data type
height = 5.8
print("\nHeight:", height)
print("Type of height:", type(height))

# String data type
name = "Rahul"
print("\nName:", name)
print("Type of name:", type(name))

# Boolean data type
is_student = True
print("\nIs Student:", is_student)
print("Type of is_student:", type(is_student))
```

```python

# Arithmetic Operators

a = 10
b = 3

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Floor Division:", a // b)
print("Modulus:", a % b)
print("Power:", a ** b)
```

```python
# Assignment Operators

x = 10
print("Initial value:", x)

x += 5
print("After += :", x)

x -= 3
print("After -= :", x)

x *= 2
print("After *= :", x)

x /= 4
print("After /= :", x)

x //= 2
print("After //= :", x)

x %= 2
print("After %= :", x)

x **= 3
print("After **= :", x)
```

```python
# Comparison Operators

a = 10
b = 20

print("a == b :", a == b)
print("a != b :", a != b)
print("a > b  :", a > b)
print("a < b  :", a < b)
print("a >= b :", a >= b)
print("a <= b :", a <= b)
```

```python
# Logical Operators

x = True
y = False

print("x and y :", x and y)
print("x or y  :", x or y)
print("not x   :", not x)
```

```python
# Membership Operators

name = "Python"

print("'P' in name :", "P" in name)
print("'z' in name :", "z" in name)

numbers = [10, 20, 30]

print("20 in numbers :", 20 in numbers)
print("50 not in numbers :", 50 not in numbers)
```

```python
# Identity Operators

a = [1, 2, 3]
b = [1, 2, 3]
c = a

print("a is b :", a is b)
print("a is c :", a is c)

print("a is not b :", a is not b)
```

```python
# Bitwise Operators

a = 5
b = 3

print("Binary of a:", bin(a))
print("Binary of b:", bin(b))

print("AND :", a & b)
print("OR  :", a | b)
print("XOR :", a ^ b)
print("NOT :", ~a)
print("Left Shift :", a << 1)
print("Right Shift:", a >> 1)
```

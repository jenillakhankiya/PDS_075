```python
import pandas as pd
import statistics
df = pd.read_csv("employee.csv")
data = df["DailyRate"].tolist()
data_sorted = sorted(data)
print("Dataset:", data_sorted)
```

```python
mean = statistics.mean(data)
print("Mean:", mean)
```

```python
median = statistics.median(data)
print("Median:", median)
```

```python
mode = statistics.mode(data)
print("Mode:", mode)
```

```python
population_variance = statistics.pvariance(data)
print("Population Variance:", population_variance)
```

```python
sample_variance = statistics.variance(data)
print("Sample Variance:", sample_variance)
```

```python
population_std = statistics.pstdev(data)
print("Population Standard Deviation:", population_std)
```

```python
sample_std = statistics.stdev(data)
print("Sample Standard Deviation:", sample_std)
```

```python
data_range = max(data) - min(data)
print("Range:", data_range)
```

```python

q1, q2, q3 = statistics.quantiles(data, n=4, method="inclusive")
iqr = q3 - q1
print("Q1:", q1)
print("Q2:", q2)
print("Q3:", q3)
print("IQR:", iqr)
```

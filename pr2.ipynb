```python
# Creating a List
fruits = ["apple", "banana", "mango"]

print("Original List:", fruits)

# Adding elements
fruits.append("orange")
print("After append():", fruits)

# Insert element at a specific position
fruits.insert(1, "grapes")
print("After insert():", fruits)

# Remove an element
fruits.remove("banana")
print("After remove():", fruits)

# Remove last element
fruits.pop()
print("After pop():", fruits)

# Sorting list
numbers = [5, 2, 8, 1, 3]
numbers.sort()
print("After sort():", numbers)

# Reverse list
numbers.reverse()
print("After reverse():", numbers)

# Find length
print("Length of list:", len(fruits))

# Count elements
print("Count of apple:", fruits.count("apple"))

# Copy list
new_list = fruits.copy()
print("Copied list:", new_list)
```

```python
# Creating a Tuple
colors = ("red", "green", "blue", "red")

print("Original Tuple:", colors)

# Access tuple elements
print("First element:", colors[0])

# Count occurrences
print("Count of red:", colors.count("red"))

# Find index
print("Index of blue:", colors.index("blue"))

# Length of tuple
print("Length of tuple:", len(colors))
```

```python
# Creating a Set
numbers = {10, 20, 30, 40}

print("Original Set:", numbers)

# Add element
numbers.add(50)
print("After add():", numbers)

# Remove element
numbers.remove(20)
print("After remove():", numbers)

# Discard element (does not give error if element is missing)
numbers.discard(100)
print("After discard():", numbers)

# Create another set
set2 = {30, 40, 50, 60}

# Union
print("Union:", numbers.union(set2))

# Intersection
print("Intersection:", numbers.intersection(set2))

# Difference
print("Difference:", numbers.difference(set2))

# Check subset
print("Is subset:", {30,40}.issubset(numbers))

# Length
print("Length of set:", len(numbers))
```

```python
# Creating a Dictionary

student = {
    "name": "Rahul",
    "age": 20,
    "course": "Python"
}

print("Original Dictionary:", student)

# Access value
print("Student Name:", student["name"])

# Add new key-value pair
student["city"] = "Surat"
print("After adding item:", student)

# Update value
student.update({"age": 21})
print("After update():", student)

# Remove item
student.pop("course")
print("After pop():", student)

# Get keys
print("Keys:", student.keys())

# Get values
print("Values:", student.values())

# Get key-value pairs
print("Items:", student.items())

# Copy dictionary
new_student = student.copy()
print("Copied Dictionary:", new_student)

# Clear dictionary
student.clear()
print("After clear():", student)
```

```python
# Creating a String

text = "Python Programming"

print("Original String:", text)

# Convert to uppercase
print("Uppercase:", text.upper())

# Convert to lowercase
print("Lowercase:", text.lower())

# Find length
print("Length:", len(text))

# Replace characters
print("Replace:", text.replace("Python", "Java"))

# Split string
print("Split:", text.split())

# Check starting characters
print("Starts with Python:", text.startswith("Python"))

# Check ending characters
print("Ends with Programming:", text.endswith("Programming"))

# Find substring
print("Index of Programming:", text.find("Programming"))

# Count characters
print("Count of 'm':", text.count("m"))

# Remove spaces
sample = "  Hello Python  "
print("Strip:", sample.strip())

# Join strings
words = ["Python", "is", "easy"]
print("Join:", " ".join(words))

# Check string type
print("Is alphabet:", "Python".isalpha())
print("Is digit:", "123".isdigit())
```
